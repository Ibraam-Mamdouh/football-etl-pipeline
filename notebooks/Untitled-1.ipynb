```python
def get_result(row):
    if row['home_score'] > row['away_score']:
        return 'H'
    elif row['home_score'] < row['away_score']:
        return 'A'
    else : 
        return 'D'
    
df['result'] = df.apply(get_result,axis=1)
```

```python
df
```

```python
df['year'] = df['date'].dt.year
df_modern = df[df['year']>=2000].copy()
```

```python
df_modern
```

```python
text_cols = ['home_team', 'away_team', 'tournament', 'city', 'country']
for col in text_cols:
    df_modern[col] = df_modern[col].str.strip()
```

```python
df_modern['month'] = df_modern['date'].dt.month
df_modern['day'] = df_modern['date'].dt.day_of_week
```

```python
df_modern['total_goals'] = df_modern['home_score'] + df_modern['away_score']
df_modern['goal_difference'] = df_modern['home_score'] - df_modern['away_score']
```

```python
df_modern.to_parquet("clean_results.parquet", index=False)
```

```python
df_modern.to_csv("clean_results.csv", index=False)
```
